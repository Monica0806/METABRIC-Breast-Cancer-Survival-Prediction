```python

import pandas as pd
import numpy as np

from pathlib import Path
from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python

PROJECT_DIR = Path(
    "/Users/monicachaganti/Desktop/Ph.D/PROJECTS/brca_metabric"
)

DATA_FILE = (
    PROJECT_DIR
    / "processed_data"
    / "metabric_survival_cohort.csv"
)

df = pd.read_csv(DATA_FILE)

print("Dataset shape:", df.shape)

display(df.head())
```

```python

column_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (
        df.isna().mean().values * 100
    ),
    "Unique_Values": df.nunique(dropna=True).values
})

display(
    column_summary.sort_values(
        by="Missing_Percentage",
        ascending=False
    ).round(2)
)
```

```python

print("Patients:", len(df))

print("\nEvent distribution:")
print(df["EVENT"].value_counts())

print("\nSurvival time summary:")
display(df["OS_MONTHS"].describe())

print("\nMissing outcome values:")
print(df[["OS_MONTHS", "EVENT"]].isna().sum())
```

```python

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["EVENT"]
)

train_df = train_df.copy()

test_df = test_df.copy()

print("Training patients:", len(train_df))

print("Test patients:", len(test_df))

print("\nTraining event distribution:")
print(train_df["EVENT"].value_counts())

print("\nTest event distribution:")
print(test_df["EVENT"].value_counts())
```

```python

print("Training duplicate rows:", train_df.duplicated().sum())

print("Test duplicate rows:", test_df.duplicated().sum())

print(
    "Overlapping row indices:",
    len(set(train_df.index) & set(test_df.index))
)

assert set(train_df.index).isdisjoint(set(test_df.index))

print("Train-test row separation verified!")
```

```python

clinical_features = [
    "AGE_AT_DIAGNOSIS",
    "TUMOR_SIZE",
    "GRADE",
    "LYMPH_NODES_EXAMINED_POSITIVE",
    "ER_STATUS",
    "PR_STATUS",
    "HER2_STATUS"
]

clinical_features_with_stage = (
    clinical_features + ["TUMOR_STAGE"]
)

outcome_columns = [
    "OS_MONTHS",
    "EVENT"
]

X_train = train_df[clinical_features].copy()

X_test = test_df[clinical_features].copy()

X_train_stage = train_df[
    clinical_features_with_stage
].copy()

X_test_stage = test_df[
    clinical_features_with_stage
].copy()

y_train = train_df[outcome_columns].copy()

y_test = test_df[outcome_columns].copy()

print("Training features:", X_train.shape)

print("Test features:", X_test.shape)

print("Training outcomes:", y_train.shape)

print("Test outcomes:", y_test.shape)
```

```python

missing_summary = pd.DataFrame({
    "Missing_Count": X_train_stage.isna().sum(),
    "Missing_Percentage": (
        X_train_stage.isna().mean() * 100
    )
})

display(
    missing_summary.sort_values(
        by="Missing_Percentage",
        ascending=False
    ).round(2)
)
```

```python

categorical_features = [
    "ER_STATUS",
    "PR_STATUS",
    "HER2_STATUS",
    "GRADE",
    "TUMOR_STAGE"
]

for column in categorical_features:
    print("\n", column)

    print(
        X_train_stage[column].value_counts(
            dropna=False
        )
    )
```

```python

followup_summary = pd.DataFrame({
    "Training": y_train["OS_MONTHS"].describe(),
    "Test": y_test["OS_MONTHS"].describe()
})

display(followup_summary.round(2))

print(
    "\nZero-duration patients in training:",
    (y_train["OS_MONTHS"] == 0).sum()
)

print(
    "Zero-duration patients in test:",
    (y_test["OS_MONTHS"] == 0).sum()
)
```

```python

valid_train_mask = (
    y_train["OS_MONTHS"].notna()
    & y_train["EVENT"].notna()
    & (y_train["OS_MONTHS"] > 0)
)

X_train_model = X_train.loc[
    valid_train_mask
].copy()

y_train_model = y_train.loc[
    valid_train_mask
].copy()

print(
    "Original training patients:",
    len(X_train)
)

print(
    "Patients available for model fitting:",
    len(X_train_model)
)

print(
    "Excluded zero-duration patients:",
    (~valid_train_mask).sum()
)

print(
    "Test patients:",
    len(X_test)
)
```

```python

numeric_features = [
    "AGE_AT_DIAGNOSIS",
    "TUMOR_SIZE",
    "LYMPH_NODES_EXAMINED_POSITIVE"
]

categorical_features = [
    "GRADE",
    "ER_STATUS",
    "PR_STATUS",
    "HER2_STATUS"
]

print("Numerical features:", numeric_features)

print("Categorical features:", categorical_features)

assert set(
    numeric_features + categorical_features
) == set(X_train_model.columns)
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ],
    remainder="drop",
    verbose_feature_names_out=False
)

print("Preprocessing pipeline created successfully!")
```

```python

X_train_processed_array = (
    preprocessor.fit_transform(X_train_model)
)

X_test_processed_array = (
    preprocessor.transform(X_test)
)

processed_feature_names = (
    preprocessor.get_feature_names_out()
)

X_train_processed = pd.DataFrame(
    X_train_processed_array,
    columns=processed_feature_names,
    index=X_train_model.index
)

X_test_processed = pd.DataFrame(
    X_test_processed_array,
    columns=processed_feature_names,
    index=X_test.index
)

print(
    "Processed training shape:",
    X_train_processed.shape
)

print(
    "Processed test shape:",
    X_test_processed.shape
)

display(X_train_processed.head())
```

```python

print(
    "Training missing values:",
    X_train_processed.isna().sum().sum()
)

print(
    "Test missing values:",
    X_test_processed.isna().sum().sum()
)

print(
    "Same feature columns:",
    X_train_processed.columns.equals(
        X_test_processed.columns
    )
)

print(
    "Training outcome alignment:",
    X_train_processed.index.equals(
        y_train_model.index
    )
)

print(
    "Test outcome alignment:",
    X_test_processed.index.equals(
        y_test.index
    )
)

assert X_train_processed.isna().sum().sum() == 0

assert X_test_processed.isna().sum().sum() == 0

assert X_train_processed.columns.equals(
    X_test_processed.columns
)

assert X_train_processed.index.equals(
    y_train_model.index
)

assert X_test_processed.index.equals(
    y_test.index
)

print("Preprocessing checks passed!")
```

```python

import joblib

PREPARED_DIR = PROJECT_DIR / "prepared_data"

PREPARED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

X_train_processed.to_csv(
    PREPARED_DIR / "X_train_processed.csv",
    index=False
)

X_test_processed.to_csv(
    PREPARED_DIR / "X_test_processed.csv",
    index=False
)

y_train_model.to_csv(
    PREPARED_DIR / "y_train.csv",
    index=False
)

y_test.to_csv(
    PREPARED_DIR / "y_test.csv",
    index=False
)

joblib.dump(
    preprocessor,
    PREPARED_DIR / "clinical_preprocessor.joblib"
)

print("Prepared datasets and preprocessor saved successfully!")
```
